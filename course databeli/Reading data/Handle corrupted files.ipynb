```python
schema_str="id INT, customer_name STRING, product STRING, quantity INT, price DOUBLE, remarks STRING, corrupted_records STRING"
df= spark.read\
.schema(schema_str)\
.format("csv")\
.option("mode","PERMISSIVE")\
.load("/Volumes/my_files/my_schema/csv_files/my_files/sales_corrupted.csv")

display(df)
```
